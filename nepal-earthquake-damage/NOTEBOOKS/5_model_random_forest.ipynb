```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import OrdinalEncoder
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score
```

# **Model: Random Forest**

Logistic Regression and a single Decision Tree are both in the mix already -- adding
a Random Forest as the third model, since it's basically an ensemble of decision trees and
a natural next step to see if averaging over many trees beats a single tuned one.

```python
X_train = pd.read_csv("../data/X_train.csv", index_col=0)
X_test = pd.read_csv("../data/X_test.csv", index_col=0)
y_train = pd.read_csv("../data/y_train.csv", index_col=0).squeeze()
y_test = pd.read_csv("../data/y_test.csv", index_col=0).squeeze()

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
```

# **Baseline**

```python
acc_baseline = y_train.value_counts(normalize=True).max()
print("Baseline Accuracy:", round(acc_baseline, 2))
```

```python
categorical_cols = X_train.select_dtypes("object").columns.tolist()
numeric_cols = X_train.select_dtypes("number").columns.tolist()

preprocessor = ColumnTransformer([
    ("cat", OrdinalEncoder(handle_unknown="use_encoded_value", unknown_value=-1), categorical_cols),
], remainder="passthrough")
```

# **Hyper parameter Tuning**

```python
# tuning n_estimators this time instead of depth -- more trees usually helps up to a
# point, then plateaus while just costing more compute
n_estimators_options = [10, 25, 50, 100, 150, 200, 300]
training_acc = []
validation_acc = []

for n in n_estimators_options:
    m = Pipeline([
        ("preprocessor", preprocessor),
        ("randomforestclassifier", RandomForestClassifier(n_estimators=n, random_state=42, n_jobs=-1))
    ])
    m.fit(X_train, y_train)
    training_acc.append(m.score(X_train, y_train))
    validation_acc.append(m.score(X_test, y_test))

print("done")
```

```python
plt.figure(figsize=(10,6))
plt.plot(n_estimators_options, training_acc, marker="o", label="training")
plt.plot(n_estimators_options, validation_acc, marker="o", label="validation")
plt.xlabel("n_estimators")
plt.ylabel("Accuracy Score")
plt.title("Validation Curve, Random Forest")
plt.legend()
plt.savefig("../images/rf_validation_curve.png", dpi=150)
plt.show()
```

```python
best_n = n_estimators_options[int(np.argmax(validation_acc))]
print("Best n_estimators:", best_n)
```

# **Build Model**

```python
model = Pipeline([
    ("preprocessor", preprocessor),
    ("randomforestclassifier", RandomForestClassifier(n_estimators=best_n, random_state=42, n_jobs=-1))
])
model.fit(X_train, y_train)
```

# **Evaluate**

```python
acc_train = accuracy_score(y_train, model.predict(X_train))
acc_test = accuracy_score(y_test, model.predict(X_test))

print("Training Accuracy:", round(acc_train, 3))
print("Test Accuracy:", round(acc_test, 3))
```

# **Communicate Results**

```python
all_features = categorical_cols + numeric_cols
importances = model.named_steps["randomforestclassifier"].feature_importances_

feat_imp = pd.Series(importances, index=all_features).sort_values()
feat_imp
```

```python
feat_imp.plot(kind="barh", color="#2E86AB", figsize=(8,6))
plt.xlabel("Gini Importance")
plt.ylabel("Feature")
plt.title("Random Forest, Feature Importance")
plt.tight_layout()
plt.savefig("../images/rf_feature_importance.png", dpi=150)
plt.show()
```

```python
with open("../data/4_model_rf_results.txt", "w") as f:
    f.write(str(acc_test))
print("saved:", acc_test)
```
